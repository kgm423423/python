```python
import os, requests
from tqdm import trange

API_KEY = "51356777-0cff4e6dbcebc3c361f0392b8"  # 🔑 여기에 본인 API 키 입력
QUERY = "cats playing"
SAVE_DIR = "data/cats2"
TOTAL = 600
PER_PAGE = 200
BASE_URL = "https://pixabay.com/api/"

os.makedirs(SAVE_DIR, exist_ok=True)

downloaded = 0
image_id = 601
max_pages = (TOTAL // PER_PAGE) + 2

for page in trange(1, max_pages, desc="📥 Pages"):
    params = {
        "key": API_KEY,
        "q": QUERY,
        "image_type": "photo",
        "per_page": PER_PAGE,
        "page": page,
    }
    res = requests.get(BASE_URL, params=params)
    try:
        data = res.json()
    except:
        print("❌ JSON 파싱 실패. 응답:", res.text[:100])
        break

    hits = data.get("hits", [])
    if not hits:
        print("⚠️ 더 이상 결과 없음. 중단합니다.")
        break

    for hit in hits:
        if downloaded >= TOTAL: break
        url = hit["largeImageURL"]
        ext = os.path.splitext(url)[-1]
        path = os.path.join(SAVE_DIR, f"cat_{image_id:04d}{ext}")
        try:
            img = requests.get(url, timeout=5).content
            with open(path, "wb") as f: f.write(img)
            downloaded += 1
            image_id += 1
        except:
            continue

print(f"✅ {downloaded}장 저장 완료.")
```

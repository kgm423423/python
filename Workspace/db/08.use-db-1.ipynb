```python
import pymysql
import pandas as pd

# 1 DB연결

connection = pymysql.connect(
    host='localhost',
    user='root',
    password='235700',
    database='employees'
)

cur = connection.cursor()
```

```python
# 2 명령 실행

def sql_get(sql):
    cur.execute(sql)
    return cur.fetchall()

# sql 작성 + 명령 객체를 사용해서 SQL 전송
tables = sql_get('''SHOW TABLES''')
for table in tables:
    print("\n\n")
    print(table[0])
    print(sql_get(f'''DESC {table[0]}'''))
    for row in sql_get(f'''SELECT * FROM {table[0]} LIMIT 10'''):
        print(row)

```

```python
# 3 명령 적용
connection.commit()
```

```python
# 4 연결 해제
cur.close()
connection.close()
```

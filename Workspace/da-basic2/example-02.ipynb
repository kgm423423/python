```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
numRange = 6
np.random.seed(0)

dice = np.array(range(1, numRange+1))
#prob = np.arange(1, numRange+1) / 21
prob = np.array([1/6] * 6)
sample = np.random.choice(dice, size=30, replace=True, p=prob)

plt.hist(sample, bins=6, range=(1, numRange+1))
plt.show()
```

```python
numRange = 45
lotto = range(1, numRange+1)
prob = [1/45] * 45

means = []
for idx in range(10000):
    numbers = np.random.choice(lotto, 6, replace=False, p=prob)
    print(numbers)
    mean = sum(numbers) / len(numbers)
    means.append(mean)

plt.hist(means, bins=50)
plt.show()
```

```python
import random

# 1등 당첨 번호 (예시로 고정)
winning_numbers = [1,2,3,4,5,6]
print("당첨 번호:", sorted(winning_numbers))

# 카운터
count = 0

while True:
    count += 1
    my_numbers = set(random.sample(range(1, 46), 6))

    if my_numbers == winning_numbers:
        print(f"\n축하합니다! {count}번 만에 1등 당첨되었습니다!")
        print("내 번호:", sorted(my_numbers))
        break
```

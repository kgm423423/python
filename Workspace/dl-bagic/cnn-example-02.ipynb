```python
import tensorflow as tf

model = tf.keras.models.load_model("models/mnist_cnn_model.h5")
```

```python
import cv2
import time
import numpy as np

cap = cv2.VideoCapture(0)


def EncodingImg(img):
    threshold, inv_img = cv2.threshold(img, 110, 255, cv2.THRESH_BINARY_INV)
    contours, hierarchy = cv2.findContours(inv_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    h, w = inv_img.shape[:2]
    try:
        for contour in contours:
            (x, y), radius = cv2.minEnclosingCircle(contour)
            if radius > 10: # 너무 작은 박스는 제거
                xs, xe = int(x-radius), int(x+radius)
                ys, ye = int(y-radius), int(y+radius)
                xs = max(xs, 0)
                ys = max(ys, 0)
                xe = min(xe, w)
                ye = min(ye, h)
                cv2.rectangle(inv_img, (xs,ys), (xe,ye), (200,0,0), 1) # 이미지에 박스 그리기
                
                roi = inv_img[ys:ye, xs:xe]

                if roi.size == 0:
                    continue

                dst = cv2.resize(roi, dsize=(50, 50), interpolation=cv2.INTER_AREA)
                dst = cv2.resize(dst, dsize=(16, 16), interpolation=cv2.INTER_AREA)

                A = np.zeros((28,28))
                A[6:-6, 6:-6] = dst  # 16x16을 28x28 중앙에 배치
                A = A.reshape(-1, 28, 28, 1) / 255.0

                p = model.predict(A)
                num = p.argmax()
                cv2.putText( inv_img, str(num), (xs, ys), cv2.FONT_HERSHEY_PLAIN, 2, (200,0,0) )
    except Exception as e:
        print("이미지 변환 오류:", e)  # 로그 출력
        raise  # 현재 예외를 다시 던짐 (e 필요 없음)

    return inv_img

if cap.isOpened():
    while True:
        ret, img = cap.read()
        if ret:
            capture_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            encod_img = EncodingImg(capture_img)
            cv2.imshow("Capture_Image", capture_img)
            cv2.imshow("Encoding_Image", encod_img)
            if cv2.waitKey(1) == 27:
                break
            # time.sleep(0.1)
            # print(type(capture_img))
        else:
            print("켑쳐 읽기 실패")
            break
else:
    print("카메라를 찾을 수 없습니다.")

cap.release()
cv2.destroyAllWindows()
```
